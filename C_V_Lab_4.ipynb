```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def show_comparison(original,sobel,prewitt,canny):
  plt.figure(figsize=(12,10))
  #Orgianl Image
  plt.subplot(2,2,1)
  plt.imshow(original,cmap='gray')
  plt.title('Original GrayScale Image')
  plt.axis('off')

  plt.subplot(2,2,2)
  plt.imshow(sobel,cmap='gray')
  plt.title('Sobel Edge Detection')
  plt.axis('off')
  plt.subplot(2,2,3)
  plt.imshow(prewitt,cmap='gray')
  plt.title('Prewitt Edge Detection')
  plt.axis('off')
  #canny EDGE Detection
  plt.subplot(2,2,4)
  plt.imshow(canny,cmap='gray')
  plt.title('Canny Edge Detection')
  plt.axis('off')

  plt.tight_layout()
  plt.show()

#Read an Image
img = cv2.imread('/content/thar.jpg',0)
if img is None:
  print('Error:image not found')
else:
  #Sobel Operation
  sobelx = cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
  sobely = cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)
  # Calculate gradient magnitude
  sobel_combined = np.sqrt(sobelx**2 + sobely**2)
  # Normalize to 0-255 and convert to uint8
  sobel_combined_display = cv2.normalize(sobel_combined, None, 0, 255, cv2.NORM_MINMAX, cv2.CV_8U)

  #Prewitt Operation
  kernelx = np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
  kernely = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
  prewittx = cv2.filter2D(img,cv2.CV_64F,kernelx)
  prewitty = cv2.filter2D(img,cv2.CV_64F,kernely)
  # Calculate gradient magnitude
  prewitt_combined = np.sqrt(prewittx**2 + prewitty**2)
  # Normalize to 0-255 and convert to uint8
  prewitt_combined_display = cv2.normalize(prewitt_combined, None, 0, 255, cv2.NORM_MINMAX, cv2.CV_8U)

  #Canny Operation
  canny_edges = cv2.Canny(img,100,200)

  #Display Results
  show_comparison(img, sobel_combined_display, prewitt_combined_display, canny_edges)
```
